```python
## In python, there is a module named 'ast' which has a method name 'literal_eval'
## The function of literal_eval is to convert the string of list of dictionaries to just list of 
## dictionaries.
```

```python
import ast
```

```python
type(extracted_movie_features.iloc[0].genres)
```

```python
## We will create a function which will give us a list of values
## Parameters: list of dictionaries, name of the key

def extract_values_from_list(list_of_dictionary, key='name'):
    # key='name'
    result = []
    
    if type(list_of_dictionary)==type('as'):
        list_of_dictionary = ast.literal_eval(list_of_dictionary)
        # print('in str')

    for item in list_of_dictionary:
        result.append(item[key])
    
    return result
```

```python
extracted_movie_features['genres_pro'] = extracted_movie_features['genres'].apply(extract_values_from_list)
```

<h3>
    Now, fetching the value of the Director will require two operations: 
</h3>
<ol>
    <li>
        First to check whether the job is Director or not.
    </li>
    <li>
        Second, to append the name of the Director to the final list.
    </li>
</ol>

```python
## Function for getting the name of the director

def fetch_name_of_director(list_of_dictionaries):

    result = []

    for item in ast.literal_eval(list_of_dictionaries):
        if item['job'] == 'Director':
            result.append(item['name'])

    return result
    
```

```python
extracted_movie_features['new_crew'] = extracted_movie_features['crew'].apply(fetch_name_of_director)
```

```python
 from sklearn.feature_extraction.text import CountVectorizer
```

```python
 cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
## Most of the time, the matrix will be sparse, and it is because, it may depend on the type of movie
## it is vectorizing.

print("Most repeating names: ", cv.get_feature_names_out())
```

#### We will now use stemming in order to remove the suffixes attached to base words.
#### This is because in vectorization, go, going, goes, are considered different, which is not good. Hence, we will apply stemming in order to get similar words to same form.

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stem(text):

    y = []
    for i in text.split():
         y.append(ps.stem(i))

    return " ".join(y)
```

```python
stem('counting the mangoes in the baskets to earn currencies of India')
```

```python
processed_movie_features['tags'] = processed_movie_features['tags'].apply(stem)
```

```python
processed_movie_features['tags']
```

```python
## Repeating the steps of count-vectorization

tag_vectors_2 = cv.fit_transform(processed_movie_features['tags']).toarray()
```

#### Now, we have got the movies in their corresponding vector form.
#### We will calculate distance of every movie's vector with every other movies' vectors.
#### More distance means less similarity.
#### We will calculate the cosine similarity (not euclidean distance).
#### In high dimensional data, Euclidean distance fails to capture similarity, hence, it is not a good measure.

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_scores = cosine_similarity(tag_vectors_2)
```

```python
print(similarity_scores)
```

```python
similarity_scores.shape
```

```python
similarity_scores[1]
```

#### Note: The diagonal values will always be 1, as similarity with of a movie with itself will be high.


```python
## Function to recommend movies.
## Given a movie, we will first find its index, and then fetch the corresponding vector and similarity score vector.
## On getting the similarity score vector, we will then return the first 5 similar movies.
```

```python
def recommend_movie(movie_name):

    movie_index = processed_movie_features[processed_movie_features['title']==movie_name].index[0]

    distances = similarity_scores[movie_index]

    # print(distances)

    ### Now, we will sort the scores in descending order such that we retain the index values of the movies

    ## We will use enumerate function

    movie_number_list = list(enumerate(distances))

    sorted_movies = sorted(movie_number_list, key=lambda x: x[1], reverse=True)[1:10] # Returning 1st 10 similar movies

    for i in sorted_movies:
        print(i[0], processed_movie_features['title'][i[0]])
    

    return
```
