<h3>
    The above is in list of dictionaries. We want to bring the whole above thing into the following format:
</h3>
<h4>
    [ 'Action', 'Adventure', 'Fantasy', 'Science Fiction' ]
</h4>

```python
## In python, there is a module named 'ast' which has a method name 'literal_eval'
## The function of literal_eval is to convert the string of list of dictionaries to just list of 
## dictionaries.
```

```python
import ast
```

```python
type(extracted_movie_features.iloc[0].genres)
```

```python
## We will create a function which will give us a list of values
## Parameters: list of dictionaries, name of the key

def extract_values_from_list(list_of_dictionary, key='name'):
    # key='name'
    result = []
    
    if type(list_of_dictionary)==type('as'):
        list_of_dictionary = ast.literal_eval(list_of_dictionary)
        # print('in str')

    for item in list_of_dictionary:
        result.append(item[key])
    
    return result
```

```python
extracted_movie_features['genres_pro'] = extracted_movie_features['genres'].apply(extract_values_from_list)
```

```python
extracted_movie_features.iloc[0].genres_pro
```

<h3>
    We will apply the same thing to the keywords column.
</h3>

```python
extracted_movie_features['new_keywords'] = extracted_movie_features['keywords'].apply(extract_values_from_list)
```

```python
extracted_movie_features['new_keywords']
```

<h3>
    We will apply same function to the cast. But, as we are interested in the first 3 actors, we will again process the list of values obtain the trim it to first 3 values.

<h3>
    Now, fetching the value of the Director will require two operations: 
</h3>
<ol>
    <li>
        First to check whether the job is Director or not.
    </li>
    <li>
        Second, to append the name of the Director to the final list.
    </li>
</ol>

```python
## Function for getting the name of the director

def fetch_name_of_director(list_of_dictionaries):

    result = []

    for item in ast.literal_eval(list_of_dictionaries):
        if item['job'] == 'Director':
            result.append(item['name'])

    return result
    
```

```python
extracted_movie_features['new_crew'] = extracted_movie_features['crew'].apply(fetch_name_of_director)
```

```python
extracted_movie_features['new_crew']
```

    Now, checking for overview of the movies.

```python
extracted_movie_features.iloc[0].overview
```

```python
## As the above data is in string, we will convert that into the list of separate strings in order
## to ease the tags creation part.

extracted_movie_features['new_overview'] = extracted_movie_features['overview'].apply(lambda x: x.split())
```

```python
extracted_movie_features['new_overview']
```

<br>
<h3>
    We will have to apply a transformation to the words processed and stored in the array form.
</h3>
<h3>
    We have to remove any spaces in the words, for e.g., New York -> NewYork.
</h3>
<h3>
    Why? Because it helps to create a better recommender system, as spaces may lead to unnecessary differentiation in the keywords. For e.g., Sam Worthington will have two tags: 'Sam', 'Worthington'; while, the Sam Mendes will also have two tags: 'Sam', 'Mendes'. Leading to unnecessary similarity between them.
</h3>
<br>

```python
extracted_movie_features['genres_pro'] = extracted_movie_features['genres_pro'].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
extracted_movie_features['genres_pro']
```

```python
extracted_movie_features['new_cast'] = extracted_movie_features['new_cast'].apply(lambda x: [i.replace(" ", "") for i in x])
extracted_movie_features['new_crew'] = extracted_movie_features['new_crew'].apply(lambda x: [i.replace(" ", "") for i in x])
extracted_movie_features['new_keywords'] = extracted_movie_features['new_keywords'].apply(lambda x: [i.replace(" ", "") for i in x])
extracted_movie_features['new_overview'] = extracted_movie_features['new_overview'].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
extracted_movie_features
```

```python
# Creating the new tags column
# Tags column is going to be the combination of genres, keywords, cast, crew

extracted_movie_features['tags'] = extracted_movie_features['genres_pro'] + extracted_movie_features['new_keywords'] + extracted_movie_features['new_cast'] + extracted_movie_features['new_crew']
```

```python
extracted_movie_features['tags']
```

```python
## Creating a new dataframe which have only the movie_id, title, and tags

processed_movie_features = extracted_movie_features[['movie_id', 'title', 'tags']]
```

```python
print(processed_movie_features.head())
```

<h3>
    Now, we will concatenate every element of the list into one whole string.
</h3>

```python
processed_movie_features['tags'] = processed_movie_features['tags'].apply(lambda x: " ".join(x))
```

```python
## Converting everything to lowercase, i.e., we are pre-processing text in order to obtain better representation

processed_movie_features['tags'] = processed_movie_features['tags'].apply(lambda x: x.lower())
```
